```python
def caught_speeding(speed, is_birthday):
    # Allow an extra 5 mph on birthdays
    if is_birthday:
        speed -= 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"
```

```python
num = int(input("Enter a number: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print(original, "is a Palindrome.")
else:
    print(original, "is not a Palindrome.")
```

```python
def find_number():
    for num in range(10000, 100000):
        s = str(num)

        d1 = int(s[0])
        d2 = int(s[1])
        d3 = int(s[2])
        d4 = int(s[3])
        d5 = int(s[4])

        if (d1 == d2 + 2 and
            d2 == d3 + 2 and
            d4 == d3 - 2 and
            d5 == d4 + 2 and
            d3 + d4 + d5 == d1 and
            d1 + d2 + d3 + d4 + d5 == 19):

            print("The required five-digit number is:", num)

find_number()
```
